```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv('/content/Day16_Student_Wellbeing_Survey.csv')

# Display the first 5 rows of the DataFrame
print("First 5 rows of the dataset:")
display(df.head())
```

### Task 1: Descriptive Statistics (Mean, Median, Mode)

```python
# Variables for descriptive statistics
variables_desc_stats = ['Weekly_Study_Hours', 'Average_Sleep_Hours', 'Daily_Screen_Time_Hours', 'Stress_Score', 'Academic_Readiness_Score']

print("Calculating Mean, Median, and Mode for selected variables:")

for col in variables_desc_stats:
    print(f"\n--- {col} ---")
    mean_val = df[col].mean()
    median_val = df[col].median()
    mode_val = df[col].mode().tolist() # .mode() can return multiple values, so convert to list

    print(f"Mean: {mean_val:.2f}")
    print(f"Median: {median_val:.2f}")
    print(f"Mode: {mode_val}")
    print(f"Interpretation: The average {col.replace('_', ' ').lower()} is {mean_val:.2f}. The middle value is {median_val:.2f}. The most frequent value(s) is/are {mode_val}.")
```

### Task 1: Descriptive Statistics (Range, Variance, Standard Deviation, Quartiles, IQR)

```python
print("Calculating Range, Variance, Standard Deviation, Q1, Q3, and IQR for selected variables:")

variabilities = {}

for col in variables_desc_stats:
    print(f"\n--- {col} ---")
    range_val = df[col].max() - df[col].min()
    variance_val = df[col].var()
    std_dev_val = df[col].std()
    q1_val = df[col].quantile(0.25)
    q3_val = df[col].quantile(0.75)
    iqr_val = q3_val - q1_val

    # Coefficient of Variation (CV) for relative variability
    cv_val = (std_dev_val / df[col].mean()) * 100 # as a percentage
    variabilities[col] = cv_val # Store CV for comparison

    print(f"Range: {range_val:.2f}")
    print(f"Variance: {variance_val:.2f}")
    print(f"Standard Deviation: {std_dev_val:.2f}")
    print(f"Q1 (25th percentile): {q1_val:.2f}")
    print(f"Q3 (75th percentile): {q3_val:.2f}")
    print(f"IQR (Interquartile Range): {iqr_val:.2f}")
    print(f"Coefficient of Variation (CV): {cv_val:.2f}%")
    print(f"Interpretation: The values for {col.replace('_', ' ').lower()} span {range_val:.2f}. The data points are, on average, {std_dev_val:.2f} units away from the mean. The middle 50% of the data falls within a range of {iqr_val:.2f}.")

# Identify variable with greatest variability (using Coefficient of Variation)
greatest_variability_var = max(variabilities, key=variabilities.get)
greatest_variability_cv = variabilities[greatest_variability_var]

print(f"\n--- Overall Variability ---")
print(f"The variable with the greatest variability is '{greatest_variability_var.replace('_', ' ').lower()}' with a Coefficient of Variation of {greatest_variability_cv:.2f}%. This indicates that {greatest_variability_var.replace('_', ' ').lower()} has the largest relative dispersion compared to its mean among the analyzed variables.")
```

### Task 2: Outlier Detection using IQR Method

```python
variables_outliers = ['Weekly_Study_Hours', 'Daily_Screen_Time_Hours', 'Commute_Time_Minutes', 'Monthly_Discretionary_Spending']

print("Identifying outliers using the IQR method:")

outliers_data = {}

for col in variables_outliers:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Identify outliers
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    num_outliers = len(outliers)

    print(f"\n--- {col} ---")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower Bound (Q1 - 1.5*IQR): {lower_bound:.2f}")
    print(f"Upper Bound (Q3 + 1.5*IQR): {upper_bound:.2f}")
    print(f"Number of outliers detected: {num_outliers}")

    if num_outliers > 0:
        print(f"Outliers (first 5 if more):\n{outliers[[col]].head()}")
    else:
        print("No outliers detected.")

    outliers_data[col] = {
        'lower_bound': lower_bound,
        'upper_bound': upper_bound,
        'outliers_df': outliers
    }
```

### Task 2: Compare Mean and Median Before and After Outlier Removal

```python
# Choosing 'Weekly_Study_Hours' for comparison
selected_variable = 'Weekly_Study_Hours'

print(f"\n--- Comparing Mean and Median for {selected_variable} ---")

# Before outlier removal
mean_before = df[selected_variable].mean()
median_before = df[selected_variable].median()
print(f"Before outlier removal:")
print(f"  Mean: {mean_before:.2f}")
print(f"  Median: {median_before:.2f}")

# After outlier removal
# Get the outliers for the selected variable from the previous step
lower_bound = outliers_data[selected_variable]['lower_bound']
upper_bound = outliers_data[selected_variable]['upper_bound']

df_no_outliers = df[(df[selected_variable] >= lower_bound) & (df[selected_variable] <= upper_bound)]

mean_after = df_no_outliers[selected_variable].mean()
median_after = df_no_outliers[selected_variable].median()
print(f"\nAfter outlier removal:")
print(f"  Mean: {mean_after:.2f}")
print(f"  Median: {median_after:.2f}")

print(f"\nInterpretation: After removing outliers from {selected_variable}, the mean changed from {mean_before:.2f} to {mean_after:.2f} and the median changed from {median_before:.2f} to {median_after:.2f}. This indicates that outliers had an impact on the central tendency, especially on the mean, pulling it towards the direction of the outliers. The median, being a more robust measure, is less affected.")
```

### Task 3: Probability Calculations

```python
# Total number of students
N = len(df)
print(f"Total number of students (N): {N}")

# Define events
# Event A = student has a Part_Time_Job of "Yes"
# Event B = student has Stress_Score ≥ 7
# Event C = student has Scholarship = "Yes"
# Event D = student exercises at least 3 days per week

# Calculate counts for each event
count_A = df[df['Part_Time_Job'] == 'Yes'].shape[0]
count_B = df[df['Stress_Score'] >= 7].shape[0]
count_C = df[df['Scholarship'] == 'Yes'].shape[0]
count_D = df[df['Exercise_Days_Per_Week'] >= 3].shape[0]

# Calculate individual probabilities
P_A = count_A / N
P_B = count_B / N
P_C = count_C / N
P_D = count_D / N

print(f"\n--- Individual Probabilities ---")
print(f"P(A) = P(Part_Time_Job = 'Yes'): {P_A:.4f}")
print(f"Interpretation: The probability that a student has a part-time job is {P_A:.2%}.")
print(f"P(B) = P(Stress_Score ≥ 7): {P_B:.4f}")
print(f"Interpretation: The probability that a student has a stress score of 7 or more is {P_B:.2%}.")
print(f"P(C) = P(Scholarship = 'Yes'): {P_C:.4f}")
print(f"Interpretation: The probability that a student has a scholarship is {P_C:.2%}.")
print(f"P(D) = P(Exercises ≥ 3 days/week): {P_D:.4f}")
print(f"Interpretation: The probability that a student exercises at least 3 days per week is {P_D:.2%}.")
```

```python
print(f"\n--- Joint and Union Probabilities ---")

# P(A and B)
# Students who have a Part_Time_Job = 'Yes' AND Stress_Score ≥ 7
count_A_and_B = df[(df['Part_Time_Job'] == 'Yes') & (df['Stress_Score'] >= 7)].shape[0]
P_A_and_B = count_A_and_B / N
print(f"P(A and B): {P_A_and_B:.4f}")
print(f"Formula: P(A and B) = (Number of students with A and B) / N")
print(f"Interpretation: The probability that a student has a part-time job AND a stress score of 7 or more is {P_A_and_B:.2%}.")

# P(A or B)
# Using the formula: P(A or B) = P(A) + P(B) - P(A and B)
P_A_or_B = P_A + P_B - P_A_and_B
print(f"\nP(A or B): {P_A_or_B:.4f}")
print(f"Formula: P(A or B) = P(A) + P(B) - P(A and B)")
print(f"Interpretation: The probability that a student has a part-time job OR a stress score of 7 or more is {P_A_or_B:.2%}.")

# Alternatively, by counting:
# count_A_or_B = df[(df['Part_Time_Job'] == 'Yes') | (df['Stress_Score'] >= 7)].shape[0]
# P_A_or_B_check = count_A_or_B / N
# print(f"P(A or B) (check by counting): {P_A_or_B_check:.4f}")
```

```python
print(f"\n--- Conditional Probabilities ---")

# P(A|B) = P(A and B) / P(B)
# Probability that a student has a part-time job GIVEN they have a stress score >= 7
P_A_given_B = P_A_and_B / P_B
print(f"P(A|B) = P(A and B) / P(B): {P_A_given_B:.4f}")
print(f"Interpretation: The probability that a student has a part-time job, given they have a stress score of 7 or more, is {P_A_given_B:.2%}.")

# P(B|A) = P(A and B) / P(A)
# Probability that a student has a stress score >= 7 GIVEN they have a part-time job
P_B_given_A = P_A_and_B / P_A
print(f"\nP(B|A) = P(A and B) / P(A): {P_B_given_A:.4f}")
print(f"Interpretation: The probability that a student has a stress score of 7 or more, given they have a part-time job, is {P_B_given_A:.2%}.")
```

```python
print(f"\n--- Mutual Exclusivity ---")

# Are Year_of_Study = 1 and Year_of_Study = 4 mutually exclusive events?
# Events are mutually exclusive if their intersection is empty, i.e., P(E1 and E2) = 0.

event_Y1 = df[df['Year_of_Study'] == 1]
event_Y4 = df[df['Year_of_Study'] == 4]

# Check for intersection
intersection_Y1_Y4 = pd.merge(event_Y1, event_Y4, how='inner', on='Student_ID')

is_mutually_exclusive = intersection_Y1_Y4.empty

print(f"Number of students in Year of Study 1: {len(event_Y1)}")
print(f"Number of students in Year of Study 4: {len(event_Y4)}")
print(f"Number of students who are in both Year of Study 1 AND Year of Study 4: {len(intersection_Y1_Y4)}")

if is_mutually_exclusive:
    print("Conclusion: Yes, the events 'Year_of_Study = 1' and 'Year_of_Study = 4' are mutually exclusive because a student cannot be in both years of study simultaneously. Their intersection is empty.")
else:
    print("Conclusion: No, the events 'Year_of_Study = 1' and 'Year_of_Study = 4' are NOT mutually exclusive.")
```

```python
print(f"\n--- Independence of Events A and B ---")

# Events A and B are independent if P(A and B) = P(A) * P(B)
product_P_A_P_B = P_A * P_B

print(f"P(A and B) (calculated earlier): {P_A_and_B:.4f}")
print(f"P(A) * P(B): {product_P_A_P_B:.4f}")

# Compare the values
# We use a small tolerance for floating point comparisons
tolerance = 0.001
if abs(P_A_and_B - product_P_A_P_B) < tolerance:
    print("Conclusion: Events A and B appear to be independent because P(A and B) is approximately equal to P(A) * P(B).")
else:
    print("Conclusion: Events A and B do NOT appear to be independent because P(A and B) is significantly different from P(A) * P(B).")
print(f"Interpretation: If P(A and B) ≈ P(A) * P(B), it suggests that having a part-time job does not significantly influence a student's stress score, and vice versa. However, if they are different, it implies a relationship between the two events.")
```

```python
print(f"\n--- Bayes' Theorem Calculation for P(A|B) ---")

# Bayes' Theorem: P(A|B) = [P(B|A) * P(A)] / P(B)

# We already have P(A), P(B), and P(B|A) from previous calculations
# P_A = P(Part_Time_Job = 'Yes')
# P_B = P(Stress_Score ≥ 7)
# P_B_given_A = P(Stress_Score ≥ 7 | Part_Time_Job = 'Yes')

# We need P(B|not A) for Bayes' Theorem as P(B) can also be written as P(B|A)P(A) + P(B|not A)P(not A)
# First, calculate P(not A)
P_not_A = 1 - P_A

# Calculate P(B|not A)
# 'not A' means student does NOT have a Part_Time_Job (Part_Time_Job = 'No')
students_not_A = df[df['Part_Time_Job'] == 'No']
count_B_given_not_A = students_not_A[students_not_A['Stress_Score'] >= 7].shape[0]

# If there are no students in 'not A', this probability is undefined or 0 if B also doesn't occur.
# We assume len(students_not_A) is not zero.
if len(students_not_A) > 0:
    P_B_given_not_A = count_B_given_not_A / len(students_not_A)
else:
    P_B_given_not_A = 0 # Or handle as an edge case if 'not A' has no occurrences

print(f"P(A): {P_A:.4f}")
print(f"P(B): {P_B:.4f}")
print(f"P(B|A): {P_B_given_A:.4f}")
print(f"P(not A): {P_not_A:.4f}")
print(f"P(B|not A): {P_B_given_not_A:.4f}")

# Applying Bayes' Theorem
# P(A|B) = [P(B|A) * P(A)] / [P(B|A) * P(A) + P(B|not A) * P(not A)]
# Note: This is an expanded form of P(B) in the denominator using the law of total probability

denominator_bayes = (P_B_given_A * P_A) + (P_B_given_not_A * P_not_A)

# Check if denominator is zero to avoid division by zero
if denominator_bayes == 0:
    P_A_given_B_bayes = 0 # Or handle as appropriate if P(B) is 0
    print("Warning: Denominator for Bayes' theorem is zero. P(B) might be 0.")
else:
    P_A_given_B_bayes = (P_B_given_A * P_A) / denominator_bayes

print(f"\nCalculated P(A|B) using Bayes' Theorem: {P_A_given_B_bayes:.4f}")
print(f"Directly calculated P(A|B): {P_A_given_B:.4f}")

# Verify agreement
if abs(P_A_given_B_bayes - P_A_given_B) < tolerance:
    print("Verification: The answer agrees with the direct conditional-probability calculation.")
else:
    print("Verification: There is a discrepancy between the Bayes' theorem calculation and the direct calculation.")
print(f"Interpretation: Bayes' Theorem provides a way to update the probability of event A given new evidence (event B). The agreement between the two methods confirms the consistency of our probability calculations.")
```

### Task 4: Normal Distribution Analysis using Academic_Readiness_Score

```python
academic_score = df['Academic_Readiness_Score']

# Calculate mean and standard deviation
mean_academic_score = academic_score.mean()
std_dev_academic_score = academic_score.std()

print(f"--- Academic Readiness Score Distribution ---")
print(f"Mean of Academic Readiness Score: {mean_academic_score:.2f}")
print(f"Standard Deviation of Academic Readiness Score: {std_dev_academic_score:.2f}")
print(f"Interpretation: The average academic readiness score is {mean_academic_score:.2f}, with scores typically deviating by {std_dev_academic_score:.2f} points from this average.")
```

```python
print(f"\n--- Z-score Calculation and Interpretation ---")

# Find the highest and lowest scores
highest_score = academic_score.max()
lowest_score = academic_score.min()

# Formula for Z-score: Z = (X - ) / σ
# Z-score for the highest score
z_score_highest = (highest_score - mean_academic_score) / std_dev_academic_score
print(f"Highest Academic Readiness Score: {highest_score:.2f}")
print(f"Z-score for the highest score: {z_score_highest:.2f}")
print(f"Interpretation: The student with the highest score ({highest_score:.2f}) is {z_score_highest:.2f} standard deviations above the mean. This indicates a very strong performance relative to the average.")

# Z-score for the lowest score
z_score_lowest = (lowest_score - mean_academic_score) / std_dev_academic_score
print(f"\nLowest Academic Readiness Score: {lowest_score:.2f}")
print(f"Z-score for the lowest score: {z_score_lowest:.2f}")
print(f"Interpretation: The student with the lowest score ({lowest_score:.2f}) is {abs(z_score_lowest):.2f} standard deviations below the mean. This indicates a very weak performance relative to the average.")
```

```python
print(f"\n--- 68-95-99.7 Empirical Rule ---")

# 1 Standard Deviation
range_1_std_lower = mean_academic_score - std_dev_academic_score
range_1_std_upper = mean_academic_score + std_dev_academic_score
print(f"Range within 1 standard deviation: ({range_1_std_lower:.2f}, {range_1_std_upper:.2f})")
print(f"Expected Percentage: 68% of students are expected to have an academic readiness score between {range_1_std_lower:.2f} and {range_1_std_upper:.2f}.")

# 2 Standard Deviations
range_2_std_lower = mean_academic_score - (2 * std_dev_academic_score)
range_2_std_upper = mean_academic_score + (2 * std_dev_academic_score)
print(f"\nRange within 2 standard deviations: ({range_2_std_lower:.2f}, {range_2_std_upper:.2f})")
print(f"Expected Percentage: 95% of students are expected to have an academic readiness score between {range_2_std_lower:.2f} and {range_2_std_upper:.2f}.")

# 3 Standard Deviations
range_3_std_lower = mean_academic_score - (3 * std_dev_academic_score)
range_3_std_upper = mean_academic_score + (3 * std_dev_academic_score)
print(f"\nRange within 3 standard deviations: ({range_3_std_lower:.2f}, {range_3_std_upper:.2f})")
print(f"Expected Percentage: 99.7% of students are expected to have an academic readiness score between {range_3_std_lower:.2f} and {range_3_std_upper:.2f}.")

print(f"Interpretation: The empirical rule provides a quick estimate of the proportion of data that falls within certain standard deviation ranges from the mean, assuming the data is approximately normally distributed. Most students' academic readiness scores fall within ±2 standard deviations from the mean.")
```
